```python
import matplotlib.pyplot as plt

month = ['1月','2月','3月','4月','5月','6月','7月','8月','9月','10月','11月','12月']
sales = [100,200,300,100,500,600,700,0,900,1000,1100,200]

# 一次性创建3行1列的子图
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 12))  # 设置画布大小

# 在ax1中画折线图
ax1.plot(month, sales, label='产品1')
ax1.legend()

# 在ax2中画柱状图
ax2.bar(month, sales, label='产品1')
ax2.legend()

# 在ax3中画饼图
ax3.pie(sales, labels=month, autopct='%1.1f%%', shadow=True, startangle=180, wedgeprops={'width':0.5})

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib import rcParams
import numpy as np
# rcParams['font.family'] = ['SimHei']
rcParams['font.family'] = ['Microsoft YaHei']
rcParams['axes.unicode_minus'] = False
df=pd.read_csv('D:\\python\\pytest1\\house_sales_extended.csv')
df=df.drop(columns=['url'])
df['area']=df['area'].str.replace('㎡',' ').astype(float)
df['price']=df['price'].str.replace('万',' ').astype(float)
df['toward']=df['toward'].astype('category')
df['unit']=df['unit'].str.replace('元/㎡',' ').astype(float)
df['year']=df['year'].str.replace('年建',' ').astype(int)
df=df[(df['area']<600)&(df['area']>20)]
Q1=df['price'].quantile(0.25)
Q3=df['price'].quantile(0.75)
IQR=Q3-Q1
low_price=Q1-1.5*IQR
high_price=Q3+1.5*IQR
df=df[(df['price']>low_price)&(df['price']<high_price)]
df['district']=df['address'].str.split('-').str[0]
df['floor_type']=df['floor'].str.split('(').str[0]
def floor_type(x):
    if '低' in x:
        return '低楼层'
    elif '中' in x:
        return '中楼层'
    elif '高' in x:
        return '高楼层'
    else:
        return '其他楼层'
df['floor_type2']=df['floor'].apply(floor_type)
df['直辖市']=df['city'].apply(lambda  x: True  if x in ['北京','天津','上海','重庆'] else False)
df['bedrooms']=df['rooms'].str.split('室').str[0].astype(int)
# df['rooms'].str.split('室').str[1].str.split('厅').str[0].astype(int)
df['livingrooms']=df['rooms'].str.extract('(\\d+)厅').astype(int)
df['building_year']=2016-df['year']
df['price_level']=pd.cut(df['price'],bins=4,labels=['低价','中价','高价','超高价'])
a=df[['price','area','unit','building_year']].corr()
a['price'].sort_values(ascending=False)[1:]
plt.figure(figsize=(5,5))
sns.heatmap(a,cmap='coolwarm',annot=True)
plt.title('房屋特征相关热力图')
plt.tight_layout()
```
